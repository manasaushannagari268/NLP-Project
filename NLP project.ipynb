```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
num_data=[]
for paragraph in data.split('\n\n'):
    for sentences in paragraph.split('.'):
        tokenized_sentences=tokenizer.texts_to_sequences([sentences])[0]
    for i in range(1,len(tokenized_sentences)):
        print(tokenized_sentences[:i+1])
```

```python
num_data = []

for paragraph in data.split('\n\n'): # splitting to paragraphs
    for sentences in paragraph.split('. '): # splitting sentences

        tokenized_sentence = tokenizer.texts_to_sequences([sentences])[0]

        for i in range(1, len(tokenized_sentence)):
            num_data.append(tokenized_sentence[:i+1])
```

```python
num_data
```

```python
max_length=max([len(i) for i in num_data])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
padded_data=pad_sequences(sequences=num_data,maxlen=max_length,padding='pre')
```

```python
padded_data
```

```python
x=padded_data[:,:-1]
y=padded_data[:,-1]
```

```python
x
```

```python
y
```

```python
from keras.utils import to_categorical
```

```python
y=to_categorical(y,num_classes=len(tokenizer.word_index)+1)
```

```python
y.shape
```

# model building

```python
from keras.models import Sequential
from keras.layers import Input,Embedding,LSTM,Dense,Bidirectional,AdditiveAttention
```

```python
x.shape
```

```python
vocab=len(tokenizer.word_index)
```

```python
model=Sequential()
model.add(Input(shape=(x.shape[1],)))  # no.of input columns
model.add(Embedding(input_dim=vocab+1,output_dim=200)) #converting each word (len(vocabulary)) into embedding to capture
model.add(Bidirectional(LSTM(150))) #adding 150 LSTM cells
model.add(Dense(vocab+1,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
max_length
```

```python
model.fit(x,y,epochs=100,validation_split=0.2)
```
